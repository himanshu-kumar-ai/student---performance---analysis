```python
# ===============================================================================
# STEP - 1 Print My First Project Information
# ===============================================================================

print("Student Performance Analysis")
print("My first Data Science project!")
```

```python
# ------------------------------------------------------------------------------
# STEP - 2: Import The Libraries
# ------------------------------------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# ------------------------------------------------------------------------------
# STEP - 3: Create Your First Dataset
# ------------------------------------------------------------------------------

data = {
    "Student": [
        "Aarav", "Diya", "Arjun", "Ananya", "Rohan",
        "Meera", "Kabir", "Isha", "Vivaan", "Sara"
    ],

    "Math": [88, 76, 95, 64, 82, 91, 70, 86, 78, 93],

    "Science": [85, 80, 92, 68, 79, 94, 73, 88, 75, 90],

    "English": [90, 78, 89, 72, 84, 87, 76, 91, 80, 95],

    "Attendance": [92, 75, 98, 65, 85, 96, 70, 88, 78, 94]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the complete dataset
print("Student Performance Dataset")
display(df)

# Verify dataset dimensions
print("Dataset shape:", df.shape)
```

```python
# ------------------------------------------------------------------------------
# STEP - 4: Inspect the data
# ------------------------------------------------------------------------------


print("First five rows:")
print(df.head())

print("\nDataset dimensions:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# ==============================================================================
# STEP - 5: Calculate average marks
# ==============================================================================


subjects = ["Math", "Science", "English"]

average_marks = df[subjects].mean()

print(average_marks)
```

```python
# ==============================================================================
# STEP - 6: Make your first Visualization
# ==============================================================================


average_marks.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```

```python
# ======================================================================
# STEP - 7: Inspect your dataset
# ======================================================================

#  7.1 Check the dataset structure


# Check dataset dimensions
print("Dataset shape:", df.shape)

# Check column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset information:")
df.info()

# Statistical summary
print("\nStatistical summary:")
print(df.describe())

```

```python
# 7.2 Check for missing values


print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# 7.3 Check for Duplicate rows


print("Number of duplicate rows:", df.duplicated().sum())
```

```python
#  7.4 Practice cleaning the data


df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
```

```python
# =======================================================================
# Step 8: Analyze student performance using statistics
# =======================================================================

# 8.1 Calculate the average marks for each subject


subjects = ["Math", "Science", "English"]

subject_average = df_clean[subjects].mean()

print("Average marks by subject:")
print(subject_average)
```

```python
# 8.2 Find the highest and lowest marks


print("Highest marks in each subject:")
print(df_clean[subjects].max())

print("\nLowest marks in each subject:")
print(df_clean[subjects].min())
```

```python
# 8.3 Calculate each student's average score


df_clean["Average_Marks"] = df_clean[subjects].mean(axis=1)

print(df_clean[["Student", "Average_Marks"]])
```

```python
# 8.4 Rank students by average marks


ranked_students = df_clean.sort_values(
    by="Average_Marks",
    ascending=False
)

print(ranked_students[["Student", "Average_Marks"]])
```

```python
# 8.5 Classify students by performance


def classify_performance(marks):
    if marks >= 80:
        return "High"
    elif marks >= 60:
        return "Medium"
    else:
        return "Low"

df_clean["Performance_Level"] = (
    df_clean["Average_Marks"].apply(classify_performance)
)

print(df_clean[["Student", "Average_Marks", "Performance_Level"]])
```

```python
# 8.6 Visualize the results

import matplotlib.pyplot as plt
import seaborn as sns

sns.barplot(
    data=df_clean,
    x="Student",
    y="Average_Marks"
)

plt.title("Average Marks of Each Student")
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```

```python
# 8.7 Write your first findings

print(df_clean["Performance_Level"].value_counts())
```

```python
# ======================================================================
# STEP - 9: Data Visualization
# ======================================================================

# 9.1 Subject-wise performance comparison


import matplotlib.pyplot as plt
import seaborn as sns

subjects = ["Math", "Science", "English"]

subject_average = df_clean[subjects].mean()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=subject_average.index,
    y=subject_average.values
)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.show()
```

```python
# 9.2 Student-wise performance comparison


plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_clean,
    x="Student",
    y="Average_Marks"
)

plt.title("Student-wise Average Performance")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.show()
```

```python
# 9.3 Score distribution (Histogram)


plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_clean,
    x="Average_Marks",
    bins=5,
    kde=True
)

plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# 9.4 Subject-wise score distribution (Box Plot)


plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_clean[subjects]
)

plt.title("Subject-wise Score Distribution")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
# 9.5 Correlation heatmap


correlation = df_clean[
    ["Math", "Science", "English",
     "Attendance", "Average_Marks"]
].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Student Performance Correlation Heatmap")

plt.show()

```

```python

# 9.6 Save Correlation Heatmap

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance Correlation Heatmap")

plt.savefig(
    "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ==============================================================================
# Step 10 — Attendance vs Academic Performance
# ==============================================================================

# ==================================================
# STEP 10.1 - Attendance Correlation
# ==================================================

attendance_correlation = df_clean[
    ["Attendance", "Average_Marks"]
].corr()

print("Attendance vs Academic Performance:")
print(attendance_correlation)
```

```python
# ==================================================
# STEP 10.2 - Scatter Plot
# ==================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="Attendance",
    y="Average_Marks",
    s=100
)

plt.title("Attendance vs Academic Performance")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.show()
```

```python
# ==================================================
# STEP 10.3 - Regression Analysis
# ==================================================

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df_clean,
    x="Attendance",
    y="Average_Marks"
)

plt.title("Attendance and Academic Performance Trend")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")

plt.show()
```

```python
# ==================================================
# STEP 10.4 - Low Attendance Students
# ==================================================

low_attendance = df_clean[
    df_clean["Attendance"] < 75
]

print("Students with attendance below 75%:")

print(
    low_attendance[
        ["Student", "Attendance", "Average_Marks"]
    ]
)
```

```python
# ==================================================
# STEP 10.5 - Attendance Classification
# ==================================================

def attendance_category(value):

    if value >= 90:
        return "High Attendance"

    elif value >= 75:
        return "Moderate Attendance"

    else:
        return "Low Attendance"


df_clean["Attendance_Group"] = (
    df_clean["Attendance"].apply(attendance_category)
)

print(
    df_clean[
        ["Student", "Attendance", "Attendance_Group"]
    ]
)
```

```python
# ==================================================
# STEP 10.6 - Group Performance Analysis
# ==================================================

group_analysis = df_clean.groupby(
    "Attendance_Group"
)["Average_Marks"].mean()

print("Average Marks by Attendance Group:")
print(group_analysis)
```

```python
# ==================================================
# STEP 10.7 - Attendance Group Comparison
# ==================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=df_clean,
    x="Attendance_Group",
    y="Average_Marks",
    order=[
        "Low Attendance",
        "Moderate Attendance",
        "High Attendance"
    ]
)

plt.title("Academic Performance by Attendance Group")
plt.xlabel("Attendance Group")
plt.ylabel("Average Marks")

plt.ylim(0, 100)

plt.show()
```

```python
# ==============================================================================
# Step 11 — Group-wise Performance Analysis
# ==============================================================================


# STEP 11.1 - Performance Classification

def classify_performance(marks):

    if marks >= 80:
        return "High"

    elif marks >= 60:
        return "Medium"

    else:
        return "Low"


df_clean["Performance_Group"] = (
    df_clean["Average_Marks"].apply(classify_performance)
)

print(df_clean[
    ["Student", "Average_Marks", "Performance_Group"]
])
```

```python

# STEP 11.2 - Count Performance Groups

group_counts = df_clean["Performance_Group"].value_counts()

print("Students in Each Performance Group:")
print(group_counts)
```

```python

# STEP 11.3 - Performance Group Chart

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="Performance_Group",
    order=["High", "Medium", "Low"]
)

plt.title("Number of Students by Performance Group")
plt.xlabel("Performance Group")
plt.ylabel("Number of Students")

plt.show()
```

```python

# STEP 11.4 - Subject-wise Group Comparison

group_subject_analysis = df_clean.groupby(
    "Performance_Group"
)[["Math", "Science", "English"]].mean()

print("Subject-wise Average Marks by Group:")
print(group_subject_analysis)
```

```python

# STEP 11.5 - Group-wise Subject Visualization

group_subject_analysis.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Subject Performance Across Student Groups")
plt.xlabel("Performance Group")
plt.ylabel("Average Marks")

plt.ylim(0, 100)
plt.legend(title="Subjects")
plt.xticks(rotation=0)

plt.show()
```

```python

# STEP 11.6 - Identify Students Needing Support

students_need_support = df_clean[
    df_clean["Performance_Group"] == "Low"
]

print("Students Needing Academic Support:")

print(
    students_need_support[
        ["Student", "Math", "Science",
         "English", "Average_Marks", "Attendance"]
    ]
)
```

```python

# STEP 11.7 - Strongest and Weakest Subject

subject_averages = df_clean[
    ["Math", "Science", "English"]
].mean()

strongest_subject = subject_averages.idxmax()
weakest_subject = subject_averages.idxmin()

print("Subject Averages:")
print(subject_averages)

print("\nStrongest Subject:", strongest_subject)
print("Weakest Subject:", weakest_subject)
```

```python

# STEP 11.8 - Performance Summary

print("STUDENT PERFORMANCE SUMMARY")
print("-" * 35)

print("Total Students:", len(df_clean))

print("\nPerformance Distribution:")
print(group_counts)

print("\nStrongest Subject:", strongest_subject)
print("Weakest Subject:", weakest_subject)

print("\nStudents Needing Support:",
      len(students_need_support))
```

```python
# ==============================================================================
# Step 12 — Data Insights and Recommendations
# ==============================================================================


# STEP 12.1 - Overall Dataset Insights

print("STUDENT PERFORMANCE INSIGHTS")
print("=" * 40)

print("Total Students:", len(df_clean))

print("\nAverage Marks:")
print(df_clean["Average_Marks"].mean())

print("\nHighest Average:")
print(df_clean["Average_Marks"].max())

print("\nLowest Average:")
print(df_clean["Average_Marks"].min())

print("\nAverage Attendance:")
print(df_clean["Attendance"].mean())
```

```python

# STEP 12.2 - Top Performer

top_student = df_clean.loc[
    df_clean["Average_Marks"].idxmax()
]

print("TOP PERFORMING STUDENT")
print("-" * 30)

print("Student:", top_student["Student"])
print("Average Marks:", top_student["Average_Marks"])
print("Attendance:", top_student["Attendance"])
```

```python

# STEP 12.3 - Lowest Performer

lowest_student = df_clean.loc[
    df_clean["Average_Marks"].idxmin()
]

print("STUDENT NEEDING MOST SUPPORT")
print("-" * 35)

print("Student:", lowest_student["Student"])
print("Average Marks:", lowest_student["Average_Marks"])
print("Attendance:", lowest_student["Attendance"])
```

```python

# STEP 12.4 - Subject Analysis

subjects = ["Math", "Science", "English"]

subject_average = df_clean[subjects].mean()

print("SUBJECT PERFORMANCE")
print(subject_average)

print("\nStrongest Subject:",
      subject_average.idxmax())

print("Subject Requiring Improvement:",
      subject_average.idxmin())
```

```python

# STEP 12.5 - Attendance Insights

attendance_avg = df_clean["Attendance"].mean()

low_attendance_students = df_clean[
    df_clean["Attendance"] < 75
]

print("Average Attendance:", attendance_avg)

print(
    "Students Below 75% Attendance:",
    len(low_attendance_students)
)

print("\nStudent Details:")

print(
    low_attendance_students[
        ["Student", "Attendance", "Average_Marks"]
    ]
)
```

```python

# STEP 12.6 - Automatic Findings

print("KEY FINDINGS")
print("=" * 40)

print(
    f"1. Average student score is "
    f"{df_clean['Average_Marks'].mean():.2f}"
)

print(
    f"2. Average attendance is "
    f"{df_clean['Attendance'].mean():.2f}%"
)

print(
    f"3. Strongest subject is "
    f"{subject_average.idxmax()}"
)

print(
    f"4. Subject needing improvement is "
    f"{subject_average.idxmin()}"
)

print(
    f"5. Students below 75% attendance: "
    f"{len(low_attendance_students)}"
)

print(
    f"6. Highest-performing student: "
    f"{top_student['Student']}"
)
```

```python

# STEP 12.7 - Recommendations

recommendations = []

if len(low_attendance_students) > 0:
    recommendations.append(
        "Provide support to students with low attendance."
    )

if subject_average.min() < 70:
    recommendations.append(
        f"Provide additional learning support in "
        f"{subject_average.idxmin()}."
    )

if df_clean["Average_Marks"].mean() < 70:
    recommendations.append(
        "Consider additional academic practice sessions."
    )

if not recommendations:
    recommendations.append(
        "Continue monitoring performance and maintaining "
        "effective learning practices."
    )

print("ACTIONABLE RECOMMENDATIONS")
print("=" * 40)

for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")
```
